## Section 1: Introduction

### 1.1 What is a Quantum Random Number Generator (QRNG)?
A **Quantum Random Number Generator (QRNG)** is an information-processing system that generates unpredictable numerical sequences by leveraging the fundamental, non-deterministic postulates of quantum mechanics. Unlike classical computers—which operate as deterministic state machines—a quantum system can be prepared in a linear superposition of orthogonal basis states. When subjected to projective measurement, the quantum state collapses into one of the basis states with a probability governed by **Born's Rule** ($P(x) = |\langle x | \psi \rangle|^2$). Because this collapse is an intrinsic property of nature rather than a consequence of hidden parameters or computational complexity, quantum randomness represents the gold standard of genuine entropy.

### 1.2 Why Randomness Matters in Computing and Science
High-entropy, unpredictable numbers are a critical substrate across modern computational disciplines:
1. **Modern Cryptography & Cybersecurity:** Cryptographic protocols (RSA, Elliptic Curve Cryptography, Diffie-Hellman key exchanges, AES session keys, initialization vectors, and nonces) fundamentally depend on unpredictable numbers. If an adversary can anticipate or compute the output of a random number generator, the security guarantees of the entire cryptographic infrastructure vanish.
2. **Monte Carlo Simulations & Scientific Computing:** Statistical physics, quantum chemistry, weather forecasting, financial derivatives modeling, and epidemiology utilize Monte Carlo methods that integrate over high-dimensional phase spaces. Correlated or periodic random sequences introduce systematic sampling biases and compromise numerical convergence.
3. **Artificial Intelligence & Stochastic Optimization:** Stochastic Gradient Descent (SGD), randomized weight initializations, Monte Carlo Tree Search (MCTS), and exploration-exploitation strategies in Reinforcement Learning depend upon uniform stochasticity.
4. **Zero-Knowledge Proofs & Blockchain Consensus:** Provably fair lotteries, cryptographic proof-of-stake systems, and randomized validator selection require publicly verifiable, tamper-proof randomness.

### 1.3 How Quantum Superposition and Measurement Generate Randomness
In the quantum circuit model, a single qubit begins in the fiducial ground state $|0\rangle$. Applying a single-qubit **Hadamard gate ($H$)** transforms the state into an equal superposition of the computational basis states $|0\rangle$ and $|1\rangle$:

$$|0\rangle \xrightarrow{H} |+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$$

When a projective measurement in the computational (Pauli-Z) basis $\{|0\rangle, |1\rangle\}$ is executed, the projection operators are:

$$M_0 = |0\rangle\langle 0|, \quad M_1 = |1\rangle\langle 1|$$

By Born's rule, the probability of obtaining bit outcome $0$ or bit outcome $1$ is:

$$P(0) = \langle + | M_0^\dagger M_0 | + \rangle = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 50\%$$

$$P(1) = \langle + | M_1^\dagger M_1 | + \rangle = \left|\frac{1}{\sqrt{2}}\right|^2 = \frac{1}{2} = 50\%$$

According to the **wavefunction collapse postulate**, the measurement causes an irreversible transition into either $|0\rangle$ or $|1\rangle$. Repeating this preparation-measurement cycle $N$ times generates a sequence of independent and identically distributed (i.i.d.) unbiased binary bits.

### 1.4 Quantum Randomness vs Classical Pseudorandomness
- **Classical Pseudorandom Number Generators (PRNGs):** Classical algorithms (such as the Mersenne Twister MT19937, Linear Congruential Generators, or PCG64) are completely deterministic. They start with an initial internal state (the seed) and apply a mathematical recurrence equation $s_{n+1} = f(s_n)$ to output numbers. While their outputs exhibit uniform statistical distributions over periods such as $2^{19937}-1$, an adversary with knowledge of the internal state can compute all future and past outputs with 100% certainty.
- **Quantum Physical Randomness:** Quantum randomness originates from irreducible physical indeterminism. No measurement prior to the interaction can predict the individual binary outcome, making it fundamentally unpredictable in principle.
- **Quantum Simulation on Classical Hardware:** In this prototype, we simulate quantum circuits using Qiskit's `AerSimulator`. A classical simulator models the quantum state vector mathematically and samples measurement outcomes using classical pseudo-stochastic algorithms. We will carefully delineate between the properties of simulated quantum circuits and physical quantum hardware.

## Section 3: Quantum Circuit Implementation

### 3.1 Circuit Design and State Evolution
We construct a quantum circuit with:
- **1 Quantum Register (1 Qubit):** $q_0$
- **1 Classical Register (1 Classical Bit):** $c_0$

The circuit sequence is:
1. **Initialization:** The qubit $q_0$ is automatically initialized to state $|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$.
2. **Hadamard Gate Application ($H$):** We apply the Hadamard operator:
   $$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$
   The transformed state is:
   $$|\psi\rangle = H |0\rangle = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}\begin{pmatrix} 1 \\ 0 \end{pmatrix} = \frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle = |+\rangle$$
3. **Computational Basis Measurement:** We perform a projective measurement in the $Z$-basis onto the classical register $c_0$. By Born's rule, the measurement projects the state into $|0\rangle$ with probability $0.5$ or $|1\rangle$ with probability $0.5$, recording `0` or `1` into $c_0$.

```python
# Construct the core 1-qubit, 1-classical-bit QRNG circuit
qrng_circuit = QuantumCircuit(1, 1, name="QRNG_Single_Qubit")

# Step 1 & 2: Apply Hadamard gate to qubit 0
qrng_circuit.h(0)

# Step 3: Measure qubit 0 into classical register 0
qrng_circuit.measure(0, 0)

# Display text circuit diagram
print("=== Quantum Random Number Generator Circuit ===")
print(qrng_circuit.draw(output='text'))
```

```python
# Render graphical circuit diagram with Matplotlib
try:
    fig = qrng_circuit.draw(output='mpl', style='iqp')
    plt.title("QRNG Circuit Architecture: Single-Qubit Hadamard Superposition & Measurement", pad=12)
    plt.show()
except Exception as e:
    print(f"Matplotlib circuit visualization fallback: {e}")
    print(qrng_circuit.draw(output='text'))
```

## Section 4: Quantum Random Bit Generation

### 4.1 Sequence Extraction with `memory=True`
To build a true random bitstream, we must retrieve the **individual sequential measurement outcome of every single shot in chronological order**, rather than merely aggregating the total histogram counts.

In Qiskit, this is achieved by specifying `memory=True` during execution with `AerSimulator.run()`. The result's `get_memory()` method returns a sequential list of measurement outcomes (e.g. `['0', '1', '1', '0', ...]`).

```python
def generate_quantum_random_bits(circuit: QuantumCircuit, num_bits: int = 10000, seed: int = None) -> list:
    """
    Executes the QRNG circuit on Qiskit's AerSimulator and returns the raw binary bitstream.
    
    Parameters:
        circuit (QuantumCircuit): The quantum circuit with Hadamard and measurement.
        num_bits (int): The total number of shots / random bits to generate.
        seed (int): Optional simulation seed for exact reproducibility.
        
    Returns:
        list: Ordered sequence of binary integers (0s and 1s).
    """
    backend = AerSimulator()
    transpiled_circuit = transpile(circuit, backend)
    
    # Execute with memory=True to record individual chronologically ordered outcomes
    job = backend.run(
        transpiled_circuit, 
        shots=num_bits, 
        memory=True, 
        seed_simulator=seed
    )
    result = job.result()
    raw_memory = result.get_memory()
    
    # Convert binary string outputs (e.g. '0', '1') into integers
    bits = [int(bit_str) for bit_str in raw_memory]
    return bits

# Set configurable sequence length
NUM_BITS = 10000
SEED_SIMULATOR = 42  # Seed for reproducible hackathon evaluation

print(f"Executing QRNG circuit on AerSimulator for {NUM_BITS:,} shots...")
quantum_bits = generate_quantum_random_bits(qrng_circuit, num_bits=NUM_BITS, seed=SEED_SIMULATOR)
print("Generation complete.")
```

```python
# Compute frequency and summary metrics
total_generated = len(quantum_bits)
count_zeros = quantum_bits.count(0)
count_ones = quantum_bits.count(1)
pct_zeros = (count_zeros / total_generated) * 100
pct_ones = (count_ones / total_generated) * 100

# First 50 bits string representation
first_50_str = "".join(str(b) for b in quantum_bits[:50])

print("=" * 65)
print("         QUANTUM RANDOM BIT GENERATION REPORT")
print("=" * 65)
print(f"First 50 Generated Bits:\n{first_50_str}")
print("-" * 65)
print(f"Total Bits Generated (N):      {total_generated:,}")
print(f"Count of Zeros ('0'):          {count_zeros:,}  ({pct_zeros:.2f}%)")
print(f"Count of Ones ('1'):           {count_ones:,}  ({pct_ones:.2f}%)")
print(f"Measurement Counts Dict:       {{'0': {count_zeros}, '1': {count_ones}}}")
print(f"Absolute Count Difference:     {abs(count_ones - count_zeros)}")
print(f"Deviation from 50.0%:          {abs(pct_ones - 50.0):.3f}%")
print("=" * 65)
```

```python
# Store in Pandas DataFrame preserving exact sequence order
df_quantum_bits = pd.DataFrame({
    'bit_index': np.arange(total_generated),
    'quantum_bit': quantum_bits
})

# Export to downloadable CSV file
csv_filename = "qrng_simulator_bits.csv"
df_quantum_bits.to_csv(csv_filename, index=False)
print(f"[Export] Saved {len(df_quantum_bits):,} quantum bits to '{csv_filename}'.")

# Display DataFrame head and tail
print("\n--- DataFrame Head (First 5 Rows) ---")
print(df_quantum_bits.head())
print("\n--- DataFrame Tail (Last 5 Rows) ---")
print(df_quantum_bits.tail())
```

## Section 5: Statistical Randomness Analysis

To rigorously assess whether our generated bitstream conforms to an unbiased Bernoulli process ($p=0.5$), we implement four core statistical tests:

### 1. Frequency (Monobit) Test
- **Purpose:** Tests whether the proportion of ones and zeros is balanced.
- **Mathematics:** For binary bits $b_i \in \{0, 1\}$, transform to $X_i = 2b_i - 1 \in \{-1, +1\}$. The test statistic is:
  $$S_N = \sum_{i=1}^N X_i = n_1 - n_0$$
  $$Z = \frac{|S_N|}{\sqrt{N}}$$
  The complementary error function yields the two-tailed p-value:
  $$p\text{-value} = \text{erfc}\left(\frac{Z}{\sqrt{2}}\right) = 2\left(1 - \Phi(Z)\right)$$
- If $p\text{-value} \ge 0.05$, we fail to reject the null hypothesis of equal zero-one frequency.

### 2. Shannon Entropy
- **Purpose:** Measures the average information content or uncertainty per symbol.
- **Mathematics:**
  $$H(X) = - \sum_{x \in \{0, 1\}} P(x) \log_2 P(x)$$
- **Theoretical Bound:** For an ideal, independent binary random source where $P(0) = P(1) = 0.5$:
  $$H_{\max} = - (0.5 \log_2 0.5 + 0.5 \log_2 0.5) = 1.000000 \text{ bit per symbol}$$

### 3. Wald-Wolfowitz Runs Test
- **Purpose:** Determines whether consecutive sequences of identical bits (runs) occur with a frequency consistent with independent trials.
- **Mathematics:** A run is an uninterrupted sequence of like bits (e.g. `111` is a run of length 3). Let $R$ be the total observed number of runs. For $n_0$ zeros and $n_1$ ones ($N = n_0 + n_1$):
  $$\mu_R = \frac{2 n_0 n_1}{N} + 1$$
  $$\sigma_R^2 = \frac{2 n_0 n_1 (2 n_0 n_1 - N)}{N^2 (N - 1)}$$
  The standardized Z-score is:
  $$Z = \frac{R - \mu_R}{\sigma_R}$$
  $$p\text{-value} = 2 \times (1 - \Phi(|Z|))$$

### 4. Chi-Square ($\chi^2$) Goodness-of-Fit Test
- **Purpose:** Evaluates whether observed counts of zeros and ones differ significantly from the expected theoretical uniform distribution ($E_0 = E_1 = N/2$).
- **Mathematics:**
  $$\chi^2 = \frac{(n_0 - N/2)^2}{N/2} + \frac{(n_1 - N/2)^2}{N/2}$$
  With degrees of freedom $k - 1 = 2 - 1 = 1$, the p-value is computed from the $\chi^2_1$ cumulative distribution function.

```python
class RandomnessAnalyzer:
    """
    Statistical evaluation battery for binary sequences.
    """
    def __init__(self, bit_sequence: list, label: str = "Sequence"):
        self.bits = np.array(bit_sequence, dtype=int)
        self.N = len(self.bits)
        self.label = label
        self.n0 = int(np.sum(self.bits == 0))
        self.n1 = int(np.sum(self.bits == 1))
        
    def frequency_monobit_test(self) -> dict:
        """NIST-style Frequency (Monobit) Test."""
        prop_ones = self.n1 / self.N
        prop_zeros = self.n0 / self.N
        s_obs = abs(self.n1 - self.n0) / np.sqrt(self.N)
        # p-value using complementary error function
        p_value = math.erfc(s_obs / np.sqrt(2.0))
        passed = p_value >= 0.05
        return {
            'test_name': 'Frequency (Monobit) Test',
            'prop_zeros': prop_zeros,
            'prop_ones': prop_ones,
            'statistic': float(s_obs),
            'p_value': float(p_value),
            'passed': passed
        }
        
    def shannon_entropy(self) -> dict:
        """Calculates Shannon binary entropy in bits per symbol."""
        p0 = self.n0 / self.N
        p1 = self.n1 / self.N
        h0 = 0.0 if p0 == 0 else -p0 * np.log2(p0)
        h1 = 0.0 if p1 == 0 else -p1 * np.log2(p1)
        entropy = h0 + h1
        return {
            'test_name': 'Shannon Binary Entropy',
            'observed_entropy': float(entropy),
            'theoretical_max': 1.0,
            'entropy_deficit': float(1.0 - entropy)
        }
        
    def wald_wolfowitz_runs_test(self) -> dict:
        """Wald-Wolfowitz Runs Test for sequence independence."""
        # Count transitions where bit[i] != bit[i-1]
        diffs = self.bits[1:] != self.bits[:-1]
        observed_runs = int(np.sum(diffs) + 1)
        
        n0, n1, N = self.n0, self.n1, self.N
        # Expected runs and variance
        exp_runs = (2.0 * n0 * n1 / N) + 1.0
        var_runs = (2.0 * n0 * n1 * (2.0 * n0 * n1 - N)) / ((N ** 2) * (N - 1))
        std_runs = np.sqrt(var_runs)
        
        z_score = (observed_runs - exp_runs) / std_runs
        p_value = 2.0 * (1.0 - stats.norm.cdf(abs(z_score)))
        passed = p_value >= 0.05
        return {
            'test_name': 'Wald-Wolfowitz Runs Test',
            'observed_runs': observed_runs,
            'expected_runs': float(exp_runs),
            'z_score': float(z_score),
            'p_value': float(p_value),
            'passed': passed
        }
        
    def chi_square_test(self) -> dict:
        """Chi-Square Goodness-of-Fit Test against discrete uniform."""
        observed = np.array([self.n0, self.n1])
        expected = np.array([self.N / 2.0, self.N / 2.0])
        chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)
        passed = p_value >= 0.05
        return {
            'test_name': 'Chi-Square Goodness-of-Fit',
            'chi2_statistic': float(chi2_stat),
            'degrees_of_freedom': 1,
            'p_value': float(p_value),
            'passed': passed
        }
        
    def run_all_tests(self) -> dict:
        return {
            'frequency': self.frequency_monobit_test(),
            'entropy': self.shannon_entropy(),
            'runs': self.wald_wolfowitz_runs_test(),
            'chisquare': self.chi_square_test()
        }
```

```python
# Execute statistical battery on Quantum Simulator bits
analyzer_quantum = RandomnessAnalyzer(quantum_bits, label="Qiskit Aer QRNG")
results_quantum = analyzer_quantum.run_all_tests()

freq = results_quantum['frequency']
ent = results_quantum['entropy']
runs = results_quantum['runs']
chi = results_quantum['chisquare']

print("=" * 70)
print("         STATISTICAL RANDOMNESS ANALYSIS: QUANTUM SIMULATOR")
print("=" * 70)
print(f"1. FREQUENCY (MONOBIT) TEST:")
print(f"   - Proportion Zeros:    {freq['prop_zeros']:.4f}")
print(f"   - Proportion Ones:     {freq['prop_ones']:.4f}")
print(f"   - S_obs Statistic:     {freq['statistic']:.4f}")
print(f"   - Two-tailed p-value:  {freq['p_value']:.4f}")
print(f"   - Result:              {'PASS (p >= 0.05)' if freq['passed'] else 'FAIL (p < 0.05)'}")
print("-" * 70)
print(f"2. SHANNON ENTROPY:")
print(f"   - Observed Entropy:    {ent['observed_entropy']:.6f} bits / symbol")
print(f"   - Theoretical Maximum: {ent['theoretical_max']:.6f} bits / symbol")
print(f"   - Entropy Deficit:     {ent['entropy_deficit']:.6e} bits")
print("-" * 70)
print(f"3. WALD-WOLFOWITZ RUNS TEST:")
print(f"   - Observed Runs:       {runs['observed_runs']:,}")
print(f"   - Expected Runs:       {runs['expected_runs']:.2f}")
print(f"   - Standardized Z-Score:{runs['z_score']:+.4f}")
print(f"   - Two-tailed p-value:  {runs['p_value']:.4f}")
print(f"   - Result:              {'PASS (p >= 0.05)' if runs['passed'] else 'FAIL (p < 0.05)'}")
print("-" * 70)
print(f"4. CHI-SQUARE GOODNESS-OF-FIT TEST:")
print(f"   - Chi2 Statistic:      {chi['chi2_statistic']:.4f} (df = 1)")
print(f"   - p-value:             {chi['p_value']:.4f}")
print(f"   - Result:              {'PASS (p >= 0.05)' if chi['passed'] else 'FAIL (p < 0.05)'}")
print("=" * 70)
```

### Statistical Interpretation and Engineering Rigor

1. **Understanding p-values ($H_0$):** Under the null hypothesis $H_0$, each bit is an independent Bernoulli trial with $p = 0.5$. With an alpha threshold of $\alpha = 0.05$, a p-value $> 0.05$ demonstrates that the sample deviations are completely consistent with random statistical fluctuations. The null hypothesis cannot be rejected.
2. **Shannon Entropy Near Unity:** The calculated entropy exceeds $0.9999$ bits per symbol, confirming that knowledge of previous bits yields negligible reduction in uncertainty.
3. **What These Tests Establish vs What They Do NOT Establish:**
   - **What they establish:** The sequence has balanced zero-one proportions, maximizes information density, and displays natural alternating run patterns without obvious clustering or periodic traps.
   - **What they do NOT establish:** Passing elementary statistical tests **does not prove cryptographic security or physical unpredictability**. Many classical pseudorandom generators (e.g. Mersenne Twister) pass these exact tests with flying colors despite being completely predictable from seed or state reconstruction. True physical quantum unpredictability requires physical quantum hardware rather than simulation.

## Section 7: Classical Pseudorandom Comparison

### 7.1 Algorithmic PRNG Implementation
To directly demonstrate the distinctions between quantum randomness and classical algorithmic pseudorandomness, we implement a classical PRNG generator using **NumPy's modern `PCG64` BitGenerator** (`numpy.random.default_rng`).

We generate an identical sequence length ($N = 10,000$ bits) using a fixed seed (`seed = 42`) for reproducibility. We then submit both sequences to our exact identical statistical analysis battery.

### Core Difference: Determinism vs Simulated Sampling vs Physical QRNG
- **Classical PRNG:** Fully deterministic. Given the seed (`42`) and the state algorithm (`PCG64`), any party on earth can reproduce this exact 10,000-bit string bit-for-bit.
- **Quantum Simulator (Qiskit Aer):** Models the mathematical statevector $|+\rangle$ and uses classical pseudo-stochastic sampling routines in C++ to generate simulated measurement shots. It demonstrates the *logical* behavior of quantum circuits, but runs atop classical hardware.
- **Physical Quantum Processor (QPU):** Measures physical physical systems (transmon qubits, trapped ions, or photonic beam splitters). Its unpredictability is rooted in fundamental quantum vacuum fluctuations and measurement collapse.

## Section 8: Optional IBM Quantum Hardware Implementation

> **Notice:** Execution on real IBM Quantum superconducting hardware is entirely optional. This prototype functions fully independently on local simulation without requiring external cloud accounts or credentials. This section demonstrates how a production workflow interfaces with physical QPUs via the modern `qiskit-ibm-runtime` API.

### 8.1 Production Security and Graceful Handling
1. **Zero Credential Hardcoding:** We never store API tokens in notebooks. Credentials are read securely via environment variables (`os.getenv("IBM_QUANTUM_TOKEN")`) or user prompt (`getpass`).
2. **Graceful Exception Fallback:** If credentials are missing, or if quantum hardware queues are unavailable or network timeouts occur, the cell catches the exception cleanly and explains the status without halting execution.
3. **Modern V2 Primitives:** Uses `SamplerV2` and `transpile` to target backend coupling maps.

```python
import os

def run_on_ibm_quantum_hardware(circuit: QuantumCircuit, shots: int = 1000) -> dict:
    """
    Attempts execution on an actual IBM Quantum backend via Qiskit Runtime Service.
    Gracefully exits with informative diagnostics if hardware access is unconfigured.
    """
    token = os.getenv("IBM_QUANTUM_TOKEN", None)
    
    try:
        from qiskit_ibm_runtime import QiskitRuntimeService, SamplerV2
        
        # Check if an account is already saved or if environment variable is present
        if token:
            service = QiskitRuntimeService(channel="ibm_quantum", token=token)
        else:
            # Try loading existing stored local configuration
            service = QiskitRuntimeService(channel="ibm_quantum")
            
        print("[IBM Quantum] Successfully authenticated.")
        # Retrieve least busy available quantum computer
        backend = service.least_busy(operational=True, simulator=False)
        print(f"[IBM Quantum] Selected physical hardware backend: {backend.name}")
        
        # Transpile circuit for hardware topology
        transpiled_qc = transpile(circuit, backend=backend, optimization_level=1)
        sampler = SamplerV2(mode=backend)
        
        print(f"[IBM Quantum] Submitting job with {shots} shots to {backend.name}...")
        job = sampler.run([transpiled_qc], shots=shots)
        print(f"[IBM Quantum] Job ID: {job.job_id()}")
        
        # Wait for hardware execution
        result = job.result()
        return {'status': 'SUCCESS', 'result': result, 'backend': backend.name}
        
    except Exception as hardware_exception:
        print("=" * 75)
        print("             OPTIONAL IBM QUANTUM HARDWARE STATUS REPORT")
        print("=" * 75)
        print("Status: Local Simulator Prototype Operational; Physical Hardware Skipped.")
        print(f"Reason: {type(hardware_exception).__name__} - {hardware_exception}")
        print("\nHardware Setup Guide (for reviewers wishing to run on physical QPUs):")
        print("  1. Register a free IBM Quantum account at: https://quantum.ibm.com/")
        print("  2. Obtain your API token from your account dashboard.")
        print("  3. In your environment, execute:")
        print("       from qiskit_ibm_runtime import QiskitRuntimeService")
        print("       QiskitRuntimeService.save_account(channel='ibm_quantum', token='YOUR_TOKEN')")
        print("  4. Re-run this cell to submit to physical superconducting transmons.")
        print("=" * 75)
        return {'status': 'SKIPPED', 'reason': str(hardware_exception)}

# Execute optional hardware test
hardware_execution_status = run_on_ibm_quantum_hardware(qrng_circuit, shots=1000)
```

### Physical Hardware Realities vs Ideal Simulators
When a QRNG circuit executes on real physical superconducting quantum processors (such as IBM Quantum Eagle or Heron processors), several physical factors emerge:
1. **Readout Asymmetry / Readout Errors:** Measurement resonators exhibit slightly different probabilities for state classification ($|0\rangle$ classified as $|1\rangle$ vs $|1\rangle$ classified as $|0\rangle$). This introduces a slight bias (e.g. $51.2\%$ vs $48.8\%$).
2. **Decoherence ($T_1$ Relaxation & $T_2$ Dephasing):** Thermal relaxation causes excited states $|1\rangle$ to decay spontaneously toward the ground state $|0\rangle$.
3. **Post-Processing Debiasing (Von Neumann Extractor):** To recover perfect uniform randomness from a slightly biased physical quantum stream, a **Von Neumann debiaser** is applied: group bits in pairs $(b_1, b_2)$. If `01`, output `0`; if `10`, output `1`; if `00` or `11`, discard. Because $P(01) = P(10) = p(1-p)$, the extracted output is mathematically guaranteed to be strictly unbiased.

## Section 9: Results and Discussion

### 9.1 Empirical Sample Size Scaling and Statistical Fluctuations
To understand how statistical variance behaves as sample size increases, we examine three orders of magnitude: $N = 100$, $N = 1,000$, and $N = 10,000$ bits.

```python
# Empirical Sample Size Sensitivity Study
sample_sizes = [100, 1000, 10000]
scaling_records = []

for size in sample_sizes:
    sample_bits = generate_quantum_random_bits(qrng_circuit, num_bits=size, seed=100 + size)
    c0 = sample_bits.count(0)
    c1 = sample_bits.count(1)
    p0 = c0 / size
    p1 = c1 / size
    dev = abs(p1 - 0.5)
    # Standard error of the sample proportion
    se_theory = 0.5 / np.sqrt(size)
    
    scaling_records.append({
        'Sample Size (N)': f"{size:,}",
        'Count 0s': c0,
        'Count 1s': c1,
        'Observed P(1)': f"{p1:.4f}",
        'Observed Deviation |P1 - 0.5|': f"{dev:.4f}",
        'Theoretical Std Error (1/2√N)': f"{se_theory:.4f}",
        'Within ±2σ Interval': 'YES' if dev <= 2 * se_theory else 'NO'
    })

df_scaling = pd.DataFrame(scaling_records)
print("=" * 90)
print("           SAMPLE SIZE SCALING AND STATISTICAL FLUCTUATIONS")
print("=" * 90)
print(df_scaling.to_string(index=False))
print("=" * 90)
```

### 9.2 Key Technical Insights
1. **$1/\sqrt{N}$ Convergence (Central Limit Theorem):** As observed in the table above, the deviation $|P(1) - 0.5|$ scales inversely with the square root of the number of shots ($\sim 1/\sqrt{N}$). For $N=100$, statistical deviations around $2\%$ to $5\%$ are standard and expected. For $N=10,000$, deviations tighten below $0.5\%$, perfectly matching Bernoulli distribution theory.
2. **Simulation vs Physical Reality:** In our Aer simulation, pseudo-stochastic algorithms draw from the statevector $\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$. While the circuit design and API integration are 100% quantum, the execution backend on a classical computer is an algorithmic approximation. True non-deterministic physical entropy requires physical QPUs or optical quantum devices.
3. **Limitations of Basic Statistical Tests:** The Frequency, Runs, Entropy, and Chi-Square tests confirm basic uniformity and absence of consecutive clustering. However, passing these tests is a necessary baseline, not a comprehensive cryptographic certification. Full verification for cryptographic use requires batteries like **NIST SP 800-22 (15 tests)** and **Dieharder**.
